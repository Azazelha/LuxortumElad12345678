```python
http.server
import socketserver

PORT = 8000  # Приклад порту, ви можете використовувати інший

class MyHandler(http.server.SimpleHTTPRequestHandler):
    def do_GET(self):
        if self.path == "/":
            self.path = "index.html"
        elif self.path == "/persona":
            self.path = "persona.html"
        # Якщо шлях не "/" і не "/persona", можна обробити його інакше,
        # або викликати метод батьківського класу для стандартної поведінки.
        # Наприклад, для стандартної обробки файлів:
        return super().do_GET() # Викликаємо метод do_GET батьківського класу

# Далі йде частина коду для запуску сервера, яка виглядає більш-менш правильно,
# але також потребує перевірки відступів після виправлення класу.

# Handler = MyHandler # Цей рядок не потрібен, якщо ви використовуєте MyHandler безпосередньо нижче
with socketserver.TCPServer(("", PORT), MyHandler) as httpd:
    print(f"Serving at port {PORT}")
    httpd.serve_forever()

```
